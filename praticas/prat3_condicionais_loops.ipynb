```python
# Solucao 1 - Par ou impar
numero = int(input("Digite um numero inteiro: "))
if numero % 2 == 0:
    print("Par")
else:
    print("Impar")
```

```python
# Solucao 2 - Positivo, negativo ou zero
numero = float(input("Digite um numero: "))
if numero > 0:
    print("Positivo")
elif numero < 0:
    print("Negativo")
else:
    print("Zero")
```

```python
# Solucao 3 - Contagem 1 a 10
for i in range(1, 11):
    print(i)
```

```python
# Solucao 4 - Soma com while
soma = 0
numero = 1
while numero != 0:
    numero = float(input("Digite um numero (0 para parar): "))
    soma = soma + numero
print("Soma total:", soma)
```

```python
# Solucao 5 - Tabuada
n = int(input("Digite um numero para ver a tabuada: "))
for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

```python
# Solucao 6 - Maior de tres
a = float(input("Primeiro numero: "))
b = float(input("Segundo numero: "))
c = float(input("Terceiro numero: "))
if a >= b and a >= c:
    print("Maior:", a)
elif b >= a and b >= c:
    print("Maior:", b)
else:
    print("Maior:", c)
```

```python
# Solucao 7 - Pares de 0 a 20
for i in range(21):
    if i % 2 == 0:
        print(i)
```
