## Exercicio 1 - Par ou impar

Peca um numero inteiro ao usuario com input(). Use if e else para imprimir "Par" se o numero for par e "Impar" se for impar. Dica: use o resto da divisao por 2 (numero % 2).

```python
numero = int(input("Digite um numero inteiro: "))

# Verifica se o número é par
if numero % 2 == 0: # '%' é o operador de módulo, que retorna o resto da divisão
    print("Par")

else:
    print("Impar")
```

## Exercicio 2 - Positivo, negativo ou zero

Peca um numero ao usuario. Use if, elif e else para imprimir "Positivo", "Negativo" ou "Zero" conforme o valor.

```python
numero = float(input("Digite um numero: "))

# Verifica se o número é positivo, negativo ou zero
if numero > 0:
    print("Positivo")
elif numero < 0:
    print("Negativo")
else:
    print("Zero")
```

## Exercicio 3 - Contagem com for

Use um loop for com range() para imprimir os numeros de 1 a 10, um por linha.

```python
# Usando o for para imprimir os numeros de 1 a 10, um por linha.
for i in range(11):

    print(i)
```

## Exercicio 4 - Soma com while

Use um loop while para pedir numeros ao usuario ate ele digitar 0. Some todos os numeros digitados e no final imprima a soma.

```python
soma = 0

# Loop para somar numeros
while True:
    numero = float(input("Digite um numero (0 para parar): "))

    # Se o numero for 0, o loop termina
    if numero == 0:
        break
    # Soma o numero digitado ao total
    soma += numero

print("Soma total:", soma)
```

## Exercicio 5 - Tabuada

Peca um numero ao usuario. Use um for para imprimir a tabuada desse numero de 1 a 10 (por exemplo: 5 x 1 = 5, 5 x 2 = 10, ...).

```python
# Pedir um numero ao usuario
n = int(input("Digite um numero para ver a tabuada: "))

# Loop para imprimir a tabuada
for i in range(1, 11):

    # Imprime a tabuada
    print(n, "x", i, "=", n * i)
```

## Exercicio 6 - Maior de tres

Peca tres numeros ao usuario. Use if e elif para descobrir qual e o maior dos tres e imprima o resultado.

```python
a = float(input("Primeiro numero: "))
b = float(input("Segundo numero: "))
c = float(input("Terceiro numero: "))

# Verifica qual e o maior numero
if a >= b and a >= c:
    print("Maior:", a)
elif b >= a and b >= c:
    print("Maior:", b)
else:
    print("Maior:", c)
```
