```python
#Aquecendo(String + If/Else)
# Validador de Senha Simples

# Armazenado a senha na variável
senha = "mariaa"
tentativa = 0
# Pedido para o usuário digitar sua senha
usuario = input("Digite a sua senha").strip()

# Verificando se a senha está correta
if usuario == senha:

    print("Senha está correta!")

else:

    print("Senha inválida!")
```

```python
# Armazenado a senha na variável
senha = "mariaa"
tentativa = 0

while tentativa < 3:

    # Pedido para o usuário digitar sua senha
    usuario = input("Digite a sua senha").strip()

    # Verificando se a senha está correta
    if usuario == senha:

        print("Senha está correta!")
        break

    else:

        tentativa += 1

        print("Senha inválido!")

# Se for mais tem 3 tentativa a conta vai ser bloqueada.
if tentativa == 3:

    print("Conta bloqueada!")
```

```python
#CONTADOR DE VOGAIS

palavra = input("Digite uma palavra: ").strip()

contador = 0 # contar a vogais.

# Percorre a palavra.
for letra in palavra:

    # Verificado se tem vogais dentro da letra.
    if letra in "aeiouAEIOU":
        contador += 1 # Achou uma vogal.

# Imprime na tela quantas vogais tem na palvra.
print(f"A Palavra contém {contador} vogal")
```

```python
# EXERCÍCIO - SOMATÓRIO COM WHILE

# Solicita um número ao usuário e converte para inteiro
N = int(input("Digite um número: "))

contador = 1 # Começa 1

soma = 0 # Acumulador que vai somar tudo

# Enquanto contador for menor ou igual a número digitado
while contador <= N:

    soma += contador # soma = soma + contador

    contador += 1 # Incrementa o contador

print(f"A soma de 1 até {N} é: {soma}")

"""

Você digita um número → exemplo: 5

O loop vai somar:
1 + 2 + 3 + 4 + 5

O contador vai crescendo a cada volta do while.
soma vai acumulando esse crescimento.
Quando contador passa do valor de N, o while para.
"""
print()
```

```python
"""
Programa: Lista de Compras Melhorada
Autor: (você)
Descrição:
    Sistema simples para gerenciar uma lista de compras com:
        1. Adicionar item
        2. Remover item
        3. Exibir lista formatada
        4. Sair do programa
"""

lista = []

while True:

    print("\n--------------- MENU DA LISTA -----------------")
    print("1. Adicionar item")
    print("2. Remover item")
    print("3. Exibir lista formatada")
    print("4. Sair do programa")
    print("---------------------------------------------")

    opcao = input("Escolha um número: ")

    if opcao == "4":
        print("Encerrando programa..... Ate mais!")
        break

    # Adicionado item
    elif opcao == "1":

        entrada = input("Digite o(s) item(ns): ").strip()

        # Verifica entrada vazia
        if entrada == "":

            print("Você não pode adicionar um item vazio!")
        
        # Verifica se o usuário digitou vários itens separados por vírgulas
        if "," in entrada:

            itens = entrada.split(",") # Separa por vírgula

            for item in itens:
                item = item.strip() # limpar espaço

                if item != "":
                    lista.append(item)

            print("Itens adicionados separadamente!")
        
        else:

            lista.append(entrada) #Adicionado item a lista
            print(f"'{item}': Foi adicionado com sucesso.")

    elif opcao == "2":

        # Caso lista estiver vazia e não estiver nada pra remover
        if not lista:
            print("A lista está vazia! Nada para remover.")
            continue

        remover_item = input("Escreva o nome do item a ser removido: ").strip()

        if remover_item in lista:

            lista.remove(remover_item) # Removendo item da lista
            print(f"'{remover_item}': Foi removido com sucesso.")

        else:
            print(f"'{remover_item}': Não na lista.")

    elif opcao == "3":

        # Caso a lista estaja vazia
        if not lista:
            print("A lista ainda está vazia.")

        else:
            # Caso tenha itens, exibe o título da lista
            print("\nSua lista de compras:")

             # Percorre a lista exibindo número e item
            # enumerate(lista, start=1) faz o contador começar em 1, não em 0
            for i, item in enumerate(lista, start=1):
                print(f"{i}. {item}")
    else:

        print("Opção inválida! Digite um número entre 1 e 4.")
```
