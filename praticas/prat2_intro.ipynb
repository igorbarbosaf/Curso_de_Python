```python
#Aquecendo(String + If/Else)
# Validador de Senha Simples

# Armazenado a senha na variável
senha = "mariaa"
tentativa = 0
# Pedido para o usuário digitar sua senha
usuario = input("Digite a sua senha").strip()

# Verificando se a senha está correta
if usuario == senha:

    print("Senha está correta!")

else:

    print("Senha inválida!")
```

```python
# Armazenado a senha na variável
senha = "mariaa"
tentativa = 0

while tentativa < 3:

    # Pedido para o usuário digitar sua senha
    usuario = input("Digite a sua senha").strip()

    # Verificando se a senha está correta
    if usuario == senha:

        print("Senha está correta!")
        break

    else:

        tentativa += 1

        print("Senha inválido!")

# Se for mais tem 3 tentativa a conta vai ser bloqueada.
if tentativa == 3:

    print("Conta bloqueada!")
```

```python
#CONTADOR DE VOGAIS

palavra = input("Digite uma palavra: ").strip()

contador = 0 # contar a vogais.

# Percorre a palavra.
for letra in palavra:

    # Verificado se tem vogais dentro da letra.
    if letra in "aeiouAEIOU":
        contador += 1 # Achou uma vogal.

# Imprime na tela quantas vogais tem na palvra.
print(f"A Palavra contém {contador} vogal")
```

```python
# EXERCÍCIO - SOMATÓRIO COM WHILE

# Solicita um número ao usuário e converte para inteiro
N = int(input("Digite um número: "))

contador = 1 # Começa 1

soma = 0 # Acumulador que vai somar tudo

# Enquanto contador for menor ou igual a número digitado
while contador <= N:

    soma += contador # soma = soma + contador

    contador += 1 # Incrementa o contador

print(f"A soma de 1 até {N} é: {soma}")

"""

Você digita um número → exemplo: 5

O loop vai somar:
1 + 2 + 3 + 4 + 5

O contador vai crescendo a cada volta do while.
soma vai acumulando esse crescimento.
Quando contador passa do valor de N, o while para.
"""
print()
```
