```python
#Aquecendo(String + If/Else)
# Validador de Senha Simples

# Armazenado a senha na variável
senha = "mariaa"
tentativa = 0
# Pedido para o usuário digitar sua senha
usuario = input("Digite a sua senha").strip()

# Verificando se a senha está correta
if usuario == senha:

    print("Senha está correta!")

else:

    print("Senha inválida!")
```

```python
# Armazenado a senha na variável
senha = "mariaa"
tentativa = 0

while tentativa < 3:

    # Pedido para o usuário digitar sua senha
    usuario = input("Digite a sua senha").strip()

    # Verificando se a senha está correta
    if usuario == senha:

        print("Senha está correta!")
        break

    else:

        tentativa += 1

        print("Senha inválido!")

# Se for mais tem 3 tentativa a conta vai ser bloqueada.
if tentativa == 3:

    print("Conta bloqueada!")
```

```python
#CONTADOR DE VOGAIS

palavra = input("Digite uma palavra: ").strip()

contador = 0 # contar a vogais.

# Percorre a palavra.
for letra in palavra:

    # Verificado se tem vogais dentro da letra.
    if letra in "aeiouAEIOU":
        contador += 1 # Achou uma vogal.

# Imprime na tela quantas vogais tem na palvra.
print(f"A Palavra contém {contador} vogal")
```

```python
# EXERCÍCIO - SOMATÓRIO COM WHILE

# Solicita um número ao usuário e converte para inteiro
N = int(input("Digite um número: "))

contador = 1 # Começa 1

soma = 0 # Acumulador que vai somar tudo

# Enquanto contador for menor ou igual a número digitado
while contador <= N:

    soma += contador # soma = soma + contador

    contador += 1 # Incrementa o contador

print(f"A soma de 1 até {N} é: {soma}")

"""

Você digita um número → exemplo: 5

O loop vai somar:
1 + 2 + 3 + 4 + 5

O contador vai crescendo a cada volta do while.
soma vai acumulando esse crescimento.
Quando contador passa do valor de N, o while para.
"""
print()
```

```python
# programa que conta quantas vogais existem em uma palavra digitada

palavra = input("Digite uma palavra: ")
vogais = "aeiouAEIOU"
contador = 0  # Contador começa em zero

# Loop for para percorrer cada letra da palavra
for letra in palavra:
    if letra in vogais:  # Verifica se a letra está na string de vogais
        contador += 1  # Soma 1 ao contador quando encontra uma vogal

print(f"Na palavra tem: {contador} vogais")
```

```python
# programa que separa letras, números e símbolos digitados pelo usuário

texto = input("Digite qualquer coisa: ")

letras = []   # lista para armazenar apenas letras
numeros = []  # lista para armazenar apenas números
simbolos = [] # lista para armazenar qualquer outro caractere

# Percorre cada caractere digitado pelo usuário
for caracter in texto:

    if caracter.isalpha():  # verifica se o caractere é uma letra
        letras.append(caracter)  # adiciona o caractere na lista de letras

    elif caracter.isdigit():  # verifica se é número
        numeros.append(caracter)  # adiciona na lista de números

    else:  # qualquer outro tipo de caractere (espaço, pontuação, etc)
        simbolos.append(caracter)  # adiciona na lista de símbolos

print("Letras:", letras)
print("Números:", numeros)
print("Símbolos:", simbolos)
```

```python
#Limpar texto e contar cada tipo de caractere
def analista_texto(texto):

    # Removendo espaço do texto
    texto = texto.strip()

    letras = 0
    numeros = 0
    simbolos = 0

    # Percorrendo a variável texto
    for c in texto:

        if c.isalpha(): # Verificando se o texto tem letras

            letras += 1 # Somado +1

        elif c.isdigit(): # Verificando se na variável texto tem número

            numeros += 1 # Somado +1

        else:

            simbolos += 1 # 

    # Retorna as variáveis 
    return letras, numeros, simbolos

entrada = input("Digite qualquer coisa: ") # Define a variável entrada

l, n, s = analista_texto(entrada) # Chamado a função

print("Letras:", l)
print("Números:", n)
print("Símbolos:", s)
```
