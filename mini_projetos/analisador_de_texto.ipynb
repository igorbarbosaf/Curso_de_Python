```python
# Função que analisa o texto
def analisador_de_texto(texto):

    # Contadores
    vogais = 0
    consoantes = 0
    numeros = 0
    espacos = 0
    simbolos = 0

    # Lista de vogais
    lista_vogais = "aeiouAEIOU"

    # Percorrendo cada caractere do texto
    for c in texto:

        if c.isspace(): # Se for espaço
            espacos += 1

        elif c.isdigit(): # Se for número
            numeros += 1

        elif c.isalpha(): # Se for letra
            if c in lista_vogais: # Se a letra estiver dentro da lista de vogais
                vogais += 1
            else:  # Se é letra mas não é vogal → consoante
                consoantes += 1

        else: # Todo o resto vira símbolo
            simbolos += 1

    return vogais, consoantes, numeros, espacos, simbolos


# Entrada do usuário
entrada = input("Digite algo: ")

# Chamando a função
v, c, n, e, s = analisador_de_texto(entrada)

# Exibindo o resultado
print("Vogais: ", v)
print("Consoantes: ", c)
print("Números: ", n)
print("Espaços: ", e)
print("Símbolos: ", s)
```
