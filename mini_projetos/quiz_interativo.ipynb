```python
# Lista que armazena todas as perguntas do quiz
# Cada elemento é um DICIONÁRIO com chaves: pergunta, opcoes, correta
perguntas = [
    {
        "pergunta": "Qual é a capital do Brasil?",
        "opcoes": {"A": "Rio de Janeiro", "B": "Brasília", "C": "São Paulo", "D": "Salvador"},
        "correta": "B"
    },
    {
        "pergunta": "Quantos continentes existem?",
        "opcoes": {"A": "5", "B": "6", "C": "7", "D": "8"},
        "correta": "C"
    },
    {
        "pergunta": "Qual linguagem estamos estudando?",
        "opcoes": {"A": "Java", "B": "Python", "C": "C++", "D": "JavaScript"},
        "correta": "B"
    },
    {
        "pergunta": "O que usamos para repetir um bloco de código?",
        "opcoes": {"A": "if", "B": "loop", "C": "for ou while", "D": "print"},
        "correta": "C"
    },
    {
        "pergunta": "Em Python, listas são definidas com quais caracteres?",
        "opcoes": {"A": "( )", "B": "{ }", "C": "[ ]", "D": "< >"},
        "correta": "C"
    }
]

# Exibe quantas perguntas temos (função len() em listas)
print(f"Quiz carregado com {len(perguntas)} perguntas!")
```

```python
def exibir_pergunta(numero, pergunta_dict):
    """
    Função que exibe uma pergunta e suas opções na tela.
    Parâmetros:
        numero (int): número da pergunta (1, 2, 3...)
        pergunta_dict (dict): dicionário com 'pergunta', 'opcoes' e 'correta'
    """
    print("\n" + "="*50)
    print(f"Pergunta {numero}: {pergunta_dict['pergunta']}")
    print("="*50)
    # Loop FOR: percorre cada opção no dicionário de opções
    for letra, texto in pergunta_dict["opcoes"].items():
        print(f"  {letra}) {texto}")
    print()


def jogar_quiz():
    """
    Função principal: percorre todas as perguntas, exibe, pede resposta
    e conta quantas o usuário acertou. Usa LISTA, FOR, IF/ELIF e variáveis.
    Retorna a pontuação final (quantidade de acertos).
    """
    # Variável para acumular os acertos
    pontuacao = 0
    total = len(perguntas)

    # Loop FOR: para cada pergunta na lista 'perguntas'
    # enumerate() retorna o índice (i) e o elemento (pergunta) - conceito de listas
    for i, pergunta in enumerate(perguntas, start=1):
        exibir_pergunta(i, pergunta)
        # INPUT: recebe a resposta do usuário (conceito de entrada de dados)
        resposta = input("Sua resposta (A, B, C ou D): ").strip().upper()

        # ESTRUTURA CONDICIONAL: verifica se a resposta está correta
        if resposta == pergunta["correta"]:
            print("Correto!")
            pontuacao += 1  # Soma 1 ao contador de acertos
        elif resposta in ["A", "B", "C", "D"]:
            # Resposta válida mas errada
            print(f"Errado! A resposta correta era: {pergunta['correta']}")
        else:
            print("Opção inválida. Conte como erro.")

    # RETURN: devolve a pontuação para quem chamou a função
    return pontuacao, total
```

```python
# Chamada da função: jogar_quiz() retorna (pontuacao, total)
print("\n" + "QUIZ - CONHECIMENTOS GERAIS ".center(50, "="))
acertos, total_perguntas = jogar_quiz()

# Cálculo da porcentagem (conceito de variáveis e operações)
porcentagem = (acertos / total_perguntas) * 100 if total_perguntas > 0 else 0

# Exibição do resultado final
print("\n" + "="*50)
print("RESULTADO FINAL")
print("="*50)
print(f"Você acertou {acertos} de {total_perguntas} perguntas.")
print(f"Porcentagem: {porcentagem:.1f}%")

# Condicionais para mensagem final (if / elif / else)
if porcentagem >= 80:
    print("\nExcelente! Você mandou bem!")
elif porcentagem >= 60:
    print("\nMuito bom! Continue praticando!")
else:
    print("\nNão desanime! Revise o conteúdo e tente novamente.")

print("\nObrigado por jogar!")
```
